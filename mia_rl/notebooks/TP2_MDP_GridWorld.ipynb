In the theory class you saw the **MDP framework** and the **Bellman equations**.

We will use a **Sutton-style** gridworld:
- Grid: **4×4**
- Terminal states: **top-left (0,0)** and **bottom-right (3,3)**
- Reward: **-1 per step** (including bumping into walls)
- Discount factor: $\gamma = 0.9$ (slightly discounted)

```python
import numpy as np
import matplotlib.pyplot as plt
from dataclasses import dataclass
from typing import Dict, Tuple, List, Optional

np.set_printoptions(precision=3, suppress=True)
```

## 1) Gridworld MDP

We define:
- **States**: cells \((row,column)\)
- **Actions**: `U, D, L, R`
- **Transition**: deterministic (attempted move; if off-grid you stay)
- **Reward**: -1 each step; terminal states end the episode

We will implement an `env.step(state, action)` that returns:
`next_state, reward, done_flag`

```python
ACTIONS = ["U", "D", "L", "R"]
ACTION_TO_DELTA = {
    "U": (-1, 0),
    "D": ( 1, 0),
    "L": ( 0,-1),
    "R": ( 0, 1),
}

@dataclass(frozen=True)
class Gridworld:
    n_rows: int = 4
    n_cols: int = 4
    terminal_states: Tuple[Tuple[int,int], ...] = ((0,0), (3,3))
    step_reward: float = -1.0

    def states(self) -> List[Tuple[int,int]]:
        return [(row,column) for row in range(self.n_rows) for column in range(self.n_cols)]

    def is_terminal(self, state: Tuple[int,int]) -> bool:
        return state in self.terminal_states

    def step(self, state: Tuple[int,int], action: str) -> Tuple[Tuple[int,int], float, bool]:
        if self.is_terminal(state):
            return state, 0.0, True  # terminal state

        delta = ACTION_TO_DELTA[action]
        next_state = (state[0] + delta[0], state[1] + delta[1])
        next_state = (max(0, min(self.n_rows - 1, next_state[0])), max(0, min(self.n_cols - 1, next_state[1])))
        reward = self.step_reward
        done_flag = self.is_terminal(next_state)

        return next_state, reward, done_flag

env = Gridworld()
env.states()[:5], env.terminal_states
```

## 2) Utilities: indexing states and initializing V

We'll store $V$ (state-value function) as a 2D array `V[r,c]`.

```python
def zeros_V(env: Gridworld) -> np.ndarray:
    return np.zeros((env.n_rows, env.n_cols), dtype=float)

def all_states(env: Gridworld) -> List[Tuple[int,int]]:
    return env.states()

V0 = zeros_V(env)
V0
```

## 3) Policy definition

We start with a **uniform random policy**:
$\pi(a\mid s)=\frac{1}{4}$ for all non-terminal states.

```python
def uniform_random_policy(env: Gridworld) -> Dict[Tuple[int,int], Dict[str, float]]:
    policy = {}
    for s in env.states():
        if env.is_terminal(s):
            policy[s] = {a: 0.0 for a in ACTIONS} #ACTIONS = ["U", "D", "L", "R"]
        else:
            policy[s] = {a: 1.0/len(ACTIONS) for a in ACTIONS}
    return policy

policy = uniform_random_policy(env)
policy[(1,1)]
```

## 4) Policy Evaluation (compute $V_{\pi}$)

Bellman expectation backup:

$
V(s) \leftarrow \sum_a \pi(a|s)\left[r(s,a) + \gamma V(s')\right]
$

1. Compute the expected update for a single state (`bellman_expectation_update`)
2. Run iterative policy evaluation (`policy_evaluation`)

```python
def bellman_expectation_update(
    env: Gridworld, V: np.ndarray, policy: Dict[Tuple[int,int], Dict[str,float]],
    state: Tuple[int,int], gamma: float
) -> float:
    """Return the updated V(s) using the Bellman expectation backup."""
    if env.is_terminal(state):
        return 0.0

    # TODO 2: compute expected value over actions under pi(a|s)
    # Hints:
    #  - loop over actions
    v_new = 0.0
    # --- YOUR CODE STARTS HERE ---
    for action, prob in policy[state].items():
        next_state, reward, done_flag = env.step(state, action)
        v_new += prob * (reward + gamma * V[next_state])
    # --- YOUR CODE ENDS HERE ---
    return v_new


def policy_evaluation(
    env: Gridworld,
    policy: Dict[Tuple[int,int], Dict[str,float]],
    gamma: float,
    theta: float = 1e-6,
    max_iters: int = 10_000,
) -> Tuple[np.ndarray, int]:
    """Iterative policy evaluation."""
    V = zeros_V(env)

    for it in range(max_iters):
        delta = 0.0

        # TODO 3: sweep all states and update V in-place (or into a copy)
        # Use delta = max(delta, |V_new - V_old|). Stop if delta < theta.
        # --- YOUR CODE STARTS HERE ---
        for state in env.states():
            v_old = V[state]
            V[state] = bellman_expectation_update(env, V, policy, state, gamma)
            delta = max(delta, abs(V[state] - v_old))
        # --- YOUR CODE ENDS HERE ---

        if delta < theta:
            return V, it + 1

    return V, max_iters
```

## 5) Value Iteration (approximate $V^{*}$)

Bellman optimality backup:

$
V(s) \leftarrow \max_a \left[r(s,a) + \gamma V(s')\right]
$

We then extract a greedy policy:
$\pi(s)=\arg\max_a\,[r(s,a)+\gamma V(s')]$

```python
def bellman_optimality_update(env: Gridworld, V: np.ndarray, s: Tuple[int,int], gamma: float) -> float:
    if env.is_terminal(s):
        return 0.0
    best = -np.inf
    for a in ACTIONS:
        ns, r, done = env.step(s, a)
        best = max(best, r + gamma * V[ns[0], ns[1]])
        # max instead of summing as we were doing before: v_new += p * (r + gamma * V[ns[0], ns[1]])
    return best

def value_iteration(
    env: Gridworld,
    gamma: float,
    theta: float = 1e-6,
    max_iters: int = 10_000,
) -> Tuple[np.ndarray, int]:
    V = zeros_V(env)
    for it in range(max_iters):
        delta = 0.0
        V_old = V.copy()
        for s in env.states():
            v_new = bellman_optimality_update(env, V_old, s, gamma)
            delta = max(delta, abs(v_new - V[s[0], s[1]]))
            V[s[0], s[1]] = v_new
        if delta < theta:
            return V, it + 1
    return V, max_iters

V_star, iters_vi = value_iteration(env, gamma=gamma, theta=1e-8)
print("Converged in iterations:", iters_vi)
V_star
```

```python
def greedy_policy_from_V(env: Gridworld, V: np.ndarray, gamma: float) -> Dict[Tuple[int,int], str]:
    pi_greedy = {}
    for s in env.states():
        if env.is_terminal(s):
            pi_greedy[s] = "·"
            continue

        # TODO 4: get the best 'a' and 'q' (q is a one-step lookahead action value)
        # Hint: q(s, a) = r(s, a) + gamma * V(s')
        # --- YOUR CODE STARTS HERE ---
        best_a = None
        best_q = float('-inf')
        for action in ACTIONS:
            next_state, reward, done_flag = env.step(s, action)
            q = reward + gamma * V[next_state]
            if q > best_q:
                best_q = q
                best_a = action
        pi_greedy[s] = best_a
        # --- YOUR CODE ENDS HERE ---
    return pi_greedy

pi_star = greedy_policy_from_V(env, V_star, gamma=gamma)
pi_star[(2,2)], pi_star[(0,1)], pi_star[(1,0)]
```

## 5.5) Compute the action-value function $Q^{\pi}$

So far, we computed **state-value functions**.
We now introduce the **action-value function**.


For a deterministic environment, the Bellman expectation equation for $Q^{\pi}$ is:

$$
Q^{\pi}(s,a) = r(s,a) + \gamma \sum_{a'} \pi(a'\mid s')\,Q^{\pi}(s',a')
$$

where $(s', r, done) = \texttt{step}(s,a)$.

We'll implement **iterative evaluation of $Q^{\pi}$** and verify:

$$
V^{\pi}(s) = \sum_a \pi(a\mid s)\,Q^{\pi}(s,a)
$$

```python
def zeros_Q(env: Gridworld) -> np.ndarray:
    # Q[r,c,a_index]
    return np.zeros((env.n_rows, env.n_cols, len(ACTIONS)), dtype=float)

def policy_evaluation_Q(
    env: Gridworld,
    pi: Dict[Tuple[int,int], Dict[str,float]],
    gamma: float,
    theta: float = 1e-6,
    max_iters: int = 10_000,
) -> Tuple[np.ndarray, int]:
    Q = zeros_Q(env)

    for it in range(max_iters):
        delta = 0.0
        Q_old = Q.copy()

        for (r,c) in env.states():
            s = (r,c)
            if env.is_terminal(s):
                Q[r,c,:] = 0.0
                continue

            for ai, a in enumerate(ACTIONS):
                ns, reward, done = env.step(s, a)
                nr, nc = ns

                # Expected over next action under pi
                exp_next = 0.0
                for aj, a2 in enumerate(ACTIONS):
                    exp_next += pi[ns][a2] * Q_old[nr, nc, aj] # 0.25 * Q[nr,nc,aj]

                q_new = reward + gamma * exp_next
                delta = max(delta, abs(q_new - Q[r,c,ai]))
                Q[r,c,ai] = q_new

        if delta < theta:
            return Q, it + 1

    return Q, max_iters

Q_pi, itq = policy_evaluation_Q(env, pi, gamma=gamma, theta=1e-8)
print("Q^pi converged in iterations:", itq)
```

## 6) Visualization

We will visualize:
- The value function as numbers inside the grid
- A policy as arrows

This is the part that makes Bellman backups “click”.

```python
ARROW = {"U":"↑", "D":"↓", "L":"←", "R":"→", "·":"·"}

def plot_grid_values_and_policy(
    env: Gridworld,
    V: np.ndarray,
    policy: Optional[Dict[Tuple[int,int], str]] = None,
    title: str = "",
    value_fmt: str = "{:.2f}",
):
    fig, ax = plt.subplots(figsize=(6,6))
    ax.set_title(title)

    # draw grid lines
    ax.set_xlim(0, env.n_cols)
    ax.set_ylim(0, env.n_rows)
    ax.set_xticks(np.arange(env.n_cols+1))
    ax.set_yticks(np.arange(env.n_rows+1))
    ax.grid(True)
    ax.invert_yaxis()  # row 0 at top
    ax.set_xticklabels([])
    ax.set_yticklabels([])

    # highlight terminal states
    for (r,c) in env.terminal_states:
        rect = plt.Rectangle((c, r), 1, 1, fill=True, alpha=0.15)
        ax.add_patch(rect)

    # put values + arrows
    for r in range(env.n_rows):
        for c in range(env.n_cols):
            s = (r,c)
            v = V[r,c]
            txt = value_fmt.format(v)
            ax.text(c+0.5, r+0.45, txt, ha="center", va="center", fontsize=12)
            if policy is not None:
                a = policy[s] if s in policy else "·"
                ax.text(c+0.5, r+0.78, ARROW.get(a,"·"), ha="center", va="center", fontsize=18)

    plt.show()

plot_grid_values_and_policy(env, V_pi, None, title="Policy evaluation: V^π (uniform random π)")
```

### Exercise B — Add a “trap” cell
Pick a cell, e.g. (1,2), and make stepping *into* it give reward -10.
How does the optimal policy change? Tip: Inherit from the Gridworld class

Compare the number of iterations needed for policy iteration vs value iteration.

```python
@dataclass(frozen=True)
class TrapGridworld(Gridworld):
    trap_state: Tuple[int, int] = (1, 2)
    trap_reward: float = -10.0

    def step(self, state: Tuple[int, int], action: str) -> Tuple[Tuple[int, int], float, bool]:
        if self.is_terminal(state):
            return state, 0.0, True

        delta = ACTION_TO_DELTA[action]
        next_state = (
            max(0, min(self.n_rows - 1, state[0] + delta[0])),
            max(0, min(self.n_cols - 1, state[1] + delta[1])),
        )
        reward = self.trap_reward if next_state == self.trap_state else self.step_reward
        done_flag = self.is_terminal(next_state)
        return next_state, reward, done_flag


def deterministic_policy_from_actions(
    env: Gridworld,
    greedy_actions: Dict[Tuple[int, int], str],
) -> Dict[Tuple[int, int], Dict[str, float]]:
    policy_det = {}
    for state in env.states():
        if env.is_terminal(state):
            policy_det[state] = {action: 0.0 for action in ACTIONS}
        else:
            chosen = greedy_actions[state]
            policy_det[state] = {action: (1.0 if action == chosen else 0.0) for action in ACTIONS}
    return policy_det


def policy_iteration(
    env: Gridworld,
    gamma: float,
    theta: float = 1e-8,
    max_outer: int = 100,
):
    policy_pi = uniform_random_policy(env)
    history = []

    for outer in range(max_outer):
        V, eval_iters = policy_evaluation(env, policy_pi, gamma=gamma, theta=theta)
        greedy_actions = greedy_policy_from_V(env, V, gamma=gamma)
        improved_policy = deterministic_policy_from_actions(env, greedy_actions)
        stable = all(policy_pi[state] == improved_policy[state] for state in env.states())
        history.append((outer + 1, eval_iters, stable))
        policy_pi = improved_policy
        if stable:
            break

    return V, policy_pi, history


trap_env = TrapGridworld(trap_state=(1, 2), trap_reward=-10.0)
V_trap_vi, it_trap_vi = value_iteration(trap_env, gamma=gamma, theta=1e-8)
pi_trap_vi = greedy_policy_from_V(trap_env, V_trap_vi, gamma=gamma)
V_trap_pi, pi_trap_pi, hist_trap_pi = policy_iteration(trap_env, gamma=gamma, theta=1e-8)

print("Trap gridworld")
print(f"  value iteration outer sweeps: {it_trap_vi}")
print(f"  policy iteration outer sweeps: {len(hist_trap_pi)}")
print(f"  policy evaluation sweeps per policy-iteration step: {[item[1] for item in hist_trap_pi]}")
print(f"  greedy action at (1,1): {pi_trap_vi[(1, 1)]}")
print(f"  greedy action at (1,3): {pi_trap_vi[(1, 3)]}")

plot_grid_values_and_policy(
    trap_env,
    V_trap_vi,
    pi_trap_vi,
    title="Trap gridworld: V* and greedy policy",
)
```

### Exercise C — Make the environment stochastic
Modify `env.step` so that with probability 0.8 the intended action happens, and with probability 0.1 you slip left/right (relative to intended direction).
What changes in $V^{*}$ and the greedy policy?


```python
SLIP_ACTIONS = {
    "U": ("L", "R"),
    "D": ("R", "L"),
    "L": ("D", "U"),
    "R": ("U", "D"),
}


@dataclass(frozen=True)
class StochasticGridworld(Gridworld):
    intended_prob: float = 0.8
    slip_prob: float = 0.1

    def transition_distribution(
        self,
        state: Tuple[int, int],
        action: str,
    ) -> List[Tuple[float, Tuple[int, int], float, bool]]:
        if self.is_terminal(state):
            return [(1.0, state, 0.0, True)]

        outcomes = [
            (self.intended_prob, action),
            (self.slip_prob, SLIP_ACTIONS[action][0]),
            (self.slip_prob, SLIP_ACTIONS[action][1]),
        ]
        aggregated = {}
        for prob, actual_action in outcomes:
            delta = ACTION_TO_DELTA[actual_action]
            next_state = (
                max(0, min(self.n_rows - 1, state[0] + delta[0])),
                max(0, min(self.n_cols - 1, state[1] + delta[1])),
            )
            reward = self.step_reward
            done_flag = self.is_terminal(next_state)
            key = (next_state, reward, done_flag)
            aggregated[key] = aggregated.get(key, 0.0) + prob

        return [(prob, next_state, reward, done_flag) for (next_state, reward, done_flag), prob in aggregated.items()]

    def step(self, state: Tuple[int, int], action: str) -> Tuple[Tuple[int, int], float, bool]:
        transitions = self.transition_distribution(state, action)
        probabilities = np.array([prob for prob, _, _, _ in transitions], dtype=float)
        chosen_index = np.random.choice(len(transitions), p=probabilities)
        _, next_state, reward, done_flag = transitions[chosen_index]
        return next_state, reward, done_flag


def bellman_expectation_update_stochastic(
    env: StochasticGridworld,
    V: np.ndarray,
    policy: Dict[Tuple[int, int], Dict[str, float]],
    state: Tuple[int, int],
    gamma: float,
) -> float:
    if env.is_terminal(state):
        return 0.0

    v_new = 0.0
    for action, action_prob in policy[state].items():
        for transition_prob, next_state, reward, done_flag in env.transition_distribution(state, action):
            v_new += action_prob * transition_prob * (reward + gamma * V[next_state])
    return v_new


def policy_evaluation_stochastic(
    env: StochasticGridworld,
    policy: Dict[Tuple[int, int], Dict[str, float]],
    gamma: float,
    theta: float = 1e-6,
    max_iters: int = 10_000,
) -> Tuple[np.ndarray, int]:
    V = zeros_V(env)
    for it in range(max_iters):
        delta = 0.0
        for state in env.states():
            v_old = V[state]
            V[state] = bellman_expectation_update_stochastic(env, V, policy, state, gamma)
            delta = max(delta, abs(V[state] - v_old))
        if delta < theta:
            return V, it + 1
    return V, max_iters


def q_value_stochastic(
    env: StochasticGridworld,
    V: np.ndarray,
    state: Tuple[int, int],
    action: str,
    gamma: float,
) -> float:
    return sum(
        transition_prob * (reward + gamma * V[next_state])
        for transition_prob, next_state, reward, done_flag in env.transition_distribution(state, action)
    )


def value_iteration_stochastic(
    env: StochasticGridworld,
    gamma: float,
    theta: float = 1e-6,
    max_iters: int = 10_000,
) -> Tuple[np.ndarray, int]:
    V = zeros_V(env)
    for it in range(max_iters):
        delta = 0.0
        V_old = V.copy()
        for state in env.states():
            if env.is_terminal(state):
                continue
            v_new = max(q_value_stochastic(env, V_old, state, action, gamma) for action in ACTIONS)
            delta = max(delta, abs(v_new - V[state]))
            V[state] = v_new
        if delta < theta:
            return V, it + 1
    return V, max_iters


def greedy_policy_from_V_stochastic(
    env: StochasticGridworld,
    V: np.ndarray,
    gamma: float,
) -> Dict[Tuple[int, int], str]:
    pi_greedy = {}
    for state in env.states():
        if env.is_terminal(state):
            pi_greedy[state] = "·"
            continue

        best_action = None
        best_value = float("-inf")
        for action in ACTIONS:
            q_value = q_value_stochastic(env, V, state, action, gamma)
            if q_value > best_value:
                best_value = q_value
                best_action = action
        pi_greedy[state] = best_action
    return pi_greedy


stochastic_env = StochasticGridworld()
V_pi_sto, iters_pi_sto = policy_evaluation_stochastic(stochastic_env, policy, gamma=gamma, theta=1e-8)
V_star_sto, iters_vi_sto = value_iteration_stochastic(stochastic_env, gamma=gamma, theta=1e-8)
pi_star_sto = greedy_policy_from_V_stochastic(stochastic_env, V_star_sto, gamma=gamma)

print("Stochastic gridworld")
print(f"  policy evaluation iterations: {iters_pi_sto}")
print(f"  value iteration iterations: {iters_vi_sto}")
print(f"  V^pi[1,1] = {V_pi_sto[1,1]:.3f}")
print(f"  V*[1,1]   = {V_star_sto[1,1]:.3f}")

plot_grid_values_and_policy(
    stochastic_env,
    V_star_sto,
    pi_star_sto,
    title="Stochastic gridworld: V* and greedy policy",
)
```
